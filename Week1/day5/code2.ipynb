## chat model

```python
import os
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

messages = [
    ("system", "You are a senior software engineer and excellent teacher."),
    ("human", "Explain LangChain with a real-world example.")
]

response = llm.invoke(messages)

print(response.content[0]["text"])
```

# Prompt Templates

 Prompt template is a way to structure the 3 input to the model. It allows you to define a template with placeholders that can be filled in with specific values at runtime. This is useful for creating dynamic prompts based on user input or other variables.

```python
from langchain_groq import ChatGroq
```

```python
llmGroq=ChatGroq(
  model="llama-3.3-70b-versatile",
  api_key=os.getenv("GROQ_API_KEY"),
  temperature=0.2,
)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple words with examples."
)
```

```python
response=llmGroq.invoke(prompt_template.format(topic="LangChain"))
```

```python
print(response.content)
```

```python
# chains for sequencing instructions
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt = ChatPromptTemplate.from_template(
    "Explain {topic} like I'm 10 years old."
)
```

```python
chain = prompt | llmGroq | StrOutputParser()
```

```python
response = chain.invoke({
    "topic": "LangChain"
})
```

```python
print(response)
```
